5. Controle de Gastos Pessoais

```python
# Programa de Controle de Gastos
print("=== CONTROLE DE GASTOS PESSOAIS ===")

# Listas para armazenar os dados
nomes_gastos = []
valores_gastos = []

# Pergunta a quantidade de gastos
while True:
    try:
        quantidade = int(input("Quantos gastos deseja registar? "))
        break
    except ValueError:
        print("Entrada inválida! Por favor, digite um número inteiro (ex: 3).")
# Loop para coletar os gastos
for i in range(quantidade):
    nome = input(f"Nome do gasto {i + 1}: ")
    valor = float(input(f"Valor do gasto {i + 1}: R$ "))
    
    # Adiciona às listas usando APPEND
    nomes_gastos.append(nome)
    valores_gastos.append(valor)

# Calcula o total dos gastos
total_gastos = sum(valores_gastos)

# Exibe os resultados
print(f"TOTAL GASTO: R$ {total_gastos:.2f}")

# Verifica se está gastando muito
if total_gastos > 500:
    print("Atenção: Está a gastar acima do limite recomendável!")
else:
    print("Gastos dentro do orçamento estipulado.")
```
